```python
from src.ingestion.loaders import load_documents
from src.ingestion.cleaner import clean_documents
from src.ingestion.chunker import chunk_documents

from src.retrieval.dense import (
    create_embedding_model,
    create_vector_store,
    dense_search,
)

from src.retrieval.bm25 import (
    create_bm25_index,
    bm25_search,
)

from src.retrieval.hybrid import hybrid_search
```

```python
documents = load_documents()

cleaned_documents = clean_documents(documents)

chunks = chunk_documents(cleaned_documents)

print("Documents:", len(documents))
print("Chunks:", len(chunks))
```

```python
for i, chunk in enumerate(chunks, start=1):
    print(
        i,
        chunk.metadata.get("chunk_id"),
    )
```

```python
embeddings = create_embedding_model()

print("Embedding model loaded successfully.")
```

```python
from src.retrieval.dense import create_vector_store

vector_store = create_vector_store(chunks , embeddings)

print("Chroma vector store loaded successfully.")
```

```python
bm25 = create_bm25_index(chunks)

print("BM25 index created successfully.")
```

```python
query = "How do I define services in Docker Compose?"

print("Query:", query)
```

```python
dense_results = dense_search(
    vector_store=vector_store,
    query=query,
    k=10,
)

for rank, document in enumerate(dense_results, start=1):

    print("=" * 80)
    print(f"Rank: {rank}")

    print("Chunk ID:", document.metadata.get("chunk_id"))
    print("Domain:", document.metadata.get("domain"))
    print("Title:", document.metadata.get("title"))

    print("Header 1:", document.metadata.get("Header 1"))
    print("Header 2:", document.metadata.get("Header 2"))

    print("\nContent:")
    print(document.page_content[:500])
```

```python
bm25_results = bm25_search(
    bm25=bm25,
    chunks=chunks,
    query=query,
    k=10,
)

for rank, document in enumerate(bm25_results, start=1):

    print("=" * 80)
    print(f"Rank: {rank}")

    print("Chunk ID:", document.metadata.get("chunk_id"))
    print("Domain:", document.metadata.get("domain"))
    print("Title:", document.metadata.get("title"))

    print("Header 1:", document.metadata.get("Header 1"))
    print("Header 2:", document.metadata.get("Header 2"))

    print("\nContent:")
    print(document.page_content[:500])
```

```python
hybrid_results = hybrid_search(
    query=query,
    vector_store=vector_store,
    bm25=bm25,
    chunks=chunks,
    dense_k=20,
    bm25_k=20,
    final_k=10,
)
```

```python
for rank, document in enumerate(hybrid_results, start=1):

    print("=" * 80)
    print(f"Rank: {rank}")

    print("Chunk ID:", document.metadata.get("chunk_id"))
    print("Domain:", document.metadata.get("domain"))
    print("Title:", document.metadata.get("title"))

    print("Header 1:", document.metadata.get("Header 1"))
    print("Header 2:", document.metadata.get("Header 2"))

    print("\nContent:")
    print(document.page_content[:500])
```

```python
print("Dense Ranking")
print("-" * 50)

for rank, document in enumerate(dense_results, start=1):
    print(
        rank,
        document.metadata["chunk_id"]
    )


print("\nBM25 Ranking")
print("-" * 50)

for rank, document in enumerate(bm25_results, start=1):
    print(
        rank,
        document.metadata["chunk_id"]
    )


print("\nHybrid Ranking")
print("-" * 50)

for rank, document in enumerate(hybrid_results, start=1):
    print(
        rank,
        document.metadata["chunk_id"]
    )
```
