```python
from src.evaluation.metrics import RetrievalMetrics
import pandas as pd
```

```python
print(f"Number of questions: {len(questions)}")
```

```python
k_values = [5, 8, 10]

evaluation_results = []

for item in questions:

    question_id = item["question_id"]
    question = item["question"]
    relevant_documents = item["relevant_documents"]

    # Retrieve enough candidates for the largest K
    retrieved_documents = rag.retrieve(
        query=question,
        final_k=20,
    )

    # Rerank candidates
    reranked_documents = rag.rerank(
        query=question,
        documents=retrieved_documents,
        top_k=10,
    )

    metrics = RetrievalMetrics(
        retrieved_documents=reranked_documents,
        relevant_documents=relevant_documents,
    )

    row = {
        "question_id": question_id,
        "question": question,
        "category": item["category"],
        "domain": item["domain"],
    }

    for k in k_values:

        result = metrics.calculate(k)

        row[f"precision@{k}"] = result[f"precision@{k}"]
        row[f"recall@{k}"] = result[f"recall@{k}"]
        row[f"hit_rate@{k}"] = result[f"hit_rate@{k}"]

    # Reciprocal Rank does not depend on K
    row["reciprocal_rank"] = metrics.reciprocal_rank()

    evaluation_results.append(row)
```

```python
evaluation_df = pd.DataFrame(evaluation_results)

evaluation_df.head()
```

```python
import json
import pandas as pd
from pathlib import Path

from src.pipeline import RAGServices
from src.evaluation.metrics import GenerationEvaluator
```

```python
QUESTIONS_FILE = Path("data/evaluation/questions.json")

with QUESTIONS_FILE.open("r", encoding="utf-8") as file:
    questions = json.load(file)

len(questions)
```

```python
df = pd.DataFrame(questions)
df.head()
```

```python
generation_df = df[
    df["category"] != "out_of_scope"
].copy()

generation_df = generation_df.reset_index(drop=True)

print(f"Total questions: {len(df)}")
print(f"In-scope questions: {len(generation_df)}")
```

```python
rag = RAGServices()
```

```python
generation_evaluator = GenerationEvaluator()
```

```python
evaluation_df = (
    generation_df
    .sample(n=10, random_state=42)
    .reset_index(drop=True)
)

evaluation_df[[
    "question_id",
    "category",
    "domain",
    "question"
]]
```

```python
def build_context(documents):
    return "\n\n".join(
        document.page_content
        for document in documents
    )
```

```python
results = []

for index, row in evaluation_df.iterrows():

    question_id = row["question_id"]
    question = row["question"]
    ground_truth = row["ground_truth"]

    print(
        f"[{index + 1}/30] "
        f"Evaluating {question_id}..."
    )

    # ==============================
    # RAG
    # ==============================

    rag_result = rag.ask(question , rerank_k=5)

    documents = rag_result["documents"]

    context = build_context(documents)

    answer = rag_result["answer"]

    # ==============================
    # Generation Evaluation
    # ==============================

    evaluation = generation_evaluator.evaluate(
        question=question,
        context=context,
        ground_truth=ground_truth,
        answer=answer,
    )

    # ==============================
    # Store Results
    # ==============================

    results.append({
        "question_id": question_id,
        "question": question,
        "category": row["category"],
        "domain": row["domain"],
        "ground_truth": ground_truth,

        "answer": answer,

        "query_used": rag_result["query_used"],
        "rewritten": rag_result["rewritten"],
        "rewrite_count": rag_result["rewrite_count"],
        "retrieval_relevant": rag_result["relevant"],

        "faithfulness": evaluation["faithfulness"],
        "answer_relevance": evaluation["answer_relevance"],
        "correctness": evaluation["correctness"],

        "faithfulness_reason": evaluation["faithfulness_reason"],
        "answer_relevance_reason": evaluation["answer_relevance_reason"],
        "correctness_reason": evaluation["correctness_reason"],
    })
```

```python
generation_df = pd.DataFrame(results)

generation_df
```

```python
generation_metrics = [
    "faithfulness",
    "answer_relevance",
    "correctness",
]

generation_df[generation_metrics].mean()
```
