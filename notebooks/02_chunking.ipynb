```python
from src.ingestion.loaders import load_documents
from src.ingestion.cleaner import clean_documents
from src.ingestion.chunker import chunk_documents

# Load
documents = load_documents()

# Clean
cleaned_documents = clean_documents(documents)

# Chunk
chunks = chunk_documents(cleaned_documents)


print("Documents:", len(documents))
print("Chunks:", len(chunks))
```

```python
for i, chunk in enumerate(chunks[:5]):

    print("=" * 80)
    print("Chunk:", i + 1)

    print("Document ID:", chunk.metadata.get("document_id"))
    print("Domain:", chunk.metadata.get("domain"))
    print("Title:", chunk.metadata.get("title"))

    print(
        "Header 1:",
        chunk.metadata.get("Header 1")
    )

    print(
        "Header 2:",
        chunk.metadata.get("Header 2")
    )

    print("Length:", len(chunk.page_content))

    print("-" * 80)

    print(chunk.page_content[:500])
```
